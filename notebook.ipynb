# Evaluation Metrics for Classification

## Demo Model

### Imports

```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction import DictVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

### Data Preparation

```python
# Source "https://www.kaggle.com/datasets/blastchar/telco-customer-churn"
df = pd.read_csv('data/data.csv')

# Formating Header
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Formating Categorical Cols
categorical_columns = list(df.dtypes[df.dtypes == 'string'].index)
for c in categorical_columns:
    df[c] = df[c].str.lower().str.replace(' ', '_')

# Series to Numeric & Fixing Missig Data
df.totalcharges = pd.to_numeric(df.totalcharges, errors='coerce')
df.totalcharges = df.totalcharges.fillna(0)

# Label to Binary 0-1
df.churn = (df.churn == 'yes').astype(int)
df.head()
```

### Validation Framework

```python
# Spliting Full Train, Train, Val and Test Datasets
df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(df_full_train, test_size=0.25, random_state=1)

# Reset Index
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)

# Spliting Labels
y_train = df_train.churn.values
y_val = df_val.churn.values
y_test = df_test.churn.values

# Removing Churn Label (Avoid Mistakes)
del df_train['churn']
del df_val['churn']
del df_test['churn']
```

### Training Model

```python
# Spliting Numerical and Categorical Cols
numerical = ['tenure', 'monthlycharges', 'totalcharges']

categorical = ['gender', 'seniorcitizen', 'partner', 'dependents', 
               'phoneservice', 'multiplelines', 'internetservice', 'onlinesecurity', 
               'onlinebackup', 'deviceprotection', 'techsupport', 'streamingtv',
               'streamingmovies', 'contract', 'paperlessbilling', 'paymentmethod']
```

```python
# Encoding Categorical Values 0-1
dv = DictVectorizer(sparse=False)
train_dict = df_train[categorical + numerical].to_dict(orient='records')
X_train = dv.fit_transform(train_dict)

# Training the model
model = LogisticRegression(max_iter=10000)
model.fit(X_train, y_train)
```

## Accuracy
- Measures how many predictions the model gets right
- **OBS**: Can be misleading with imbalanced datasets.
- **Example**: Imagine 1,000 customers:
    - 950 will NOT churn (95%).
    - 50 WILL churn (5%).
- A model predicting NO churn for everyone gets 95% accuracy but misses all churners.

```python
# Encoding Categorical Values 0-1
val_dict = df_val[categorical + numerical].to_dict(orient='records')
X_val = dv.transform(val_dict)

# Model Probabilities
y_pred = model.predict_proba(X_val)[:, 1]

# Measuring Accuracy
churn_decision = (y_pred >= 0.5)
accuracy = (y_val == churn_decision).mean()

print(f'Accuracy -> {accuracy:.2%}')
```

```python
# Showing Best Accuracy by Threshold
thresholds = np.linspace(0, 1, 11)

scores=[]
for t in thresholds:
    # Manual Way
    #churn_decision = (y_pred >= t)
    #score = (y_val == churn_decision).mean()

    # Using Sklearn
    score = accuracy_score(y_val, y_pred >= t)
    print('%.2f %.3f' % (t, score))
    scores.append(score)
```

```python
# Plot Visualization
plt.plot(thresholds, scores)
```
